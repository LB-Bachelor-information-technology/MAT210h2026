## Oppgave 1

Gitt funksjonen:


$$f(x,y) = -x^3 + x^2 + x + y^3 - 3y$$

### a) Første- og andreordens partiellderiverte

Vi finner de partiellderiverte ved å derivere mhp. én variabel om gangen mens den andre behandles som en konstant:

**Førsteordens partiellderiverte:**

$$f_x(x,y) = \frac{\partial f}{\partial x} = -3x^2 + 2x + 1$$
$$f_y(x,y) = \frac{\partial f}{\partial y} = 3y^2 - 3$$

**Andreordens partiellderiverte:**

$$f_{xx}(x,y) = \frac{\partial^2 f}{\partial x^2} = -6x + 2$$
$$f_{yy}(x,y) = \frac{\partial^2 f}{\partial y^2} = 6y$$
$$f_{xy}(x,y) = f_{yx}(x,y) = \frac{\partial^2 f}{\partial x \partial y} = 0$$


```python
import sympy as sp

x, y = sp.symbols('x y')
f = -x**3 + x**2 + x + y**3 - 3*y

# Førsteordens partiellderiverte
fx = sp.diff(f, x)
fy = sp.diff(f, y)

# Andreordens partiellderiverte
fxx = sp.diff(fx, x)
fyy = sp.diff(fy, y)
fxy = sp.diff(fx, y)

print("fx  =", fx)
print("fy  =", fy)
print("fxx =", fxx)
print("fyy =", fyy)
print("fxy =", fxy)
```

### b) Retningen $f$ vokser raskest i $(-1, 1)$

Retningen der en funksjon vokser raskest er gitt ved gradientvektoren $\nabla f(x,y) = \begin{pmatrix} f_x \\ f_y \end{pmatrix}$ evaluert i punktet.

I punktet $(-1, 1)$:


$$f_x(-1,1) = -3(-1)^2 + 2(-1) + 1 = -3 - 2 + 1 = -4$$

$$f_y(-1,1) = 3(1)^2 - 3 = 0$$

Gradientvektoren er altså:


$$\nabla f(-1,1) = \begin{pmatrix} -4 \\ 0 \end{pmatrix}$$

Funksjonen vokser raskest i retning av vektoren $\begin{pmatrix} -4 \\ 0 \end{pmatrix}$ (eller den normerte enhetsvektoren $\begin{pmatrix} -1 \\ 0 \end{pmatrix}$).

```python
grad_f = sp.Matrix([fx, fy])
grad_in_pt = grad_f.subs({x: -1, y: 1})

print("Gradient i (-1, 1):", grad_in_pt)
```

### c) Finn og klassifiser de stasjonære/kritiske punktene

For å finne kritiske punkt løser vi $\nabla f(x,y) = \mathbf{0}$:

1. $f_x = -3x^2 + 2x + 1 = 0 \implies (3x+1)(1-x) = 0 \implies x = 1 \text{ eller } x = -\frac{1}{3}$
2. $f_y = 3y^2 - 3 = 0 \implies y^2 = 1 \implies y = 1 \text{ eller } y = -1$

Dette gir 4 stasjonære punkt:


$$P_1 = (1, 1), \quad P_2 = (1, -1), \quad P_3 = \left(-\frac{1}{3}, 1\right), \quad P_4 = \left(-\frac{1}{3}, -1\right)$$

For å klassifisere bruker vi Hesse-determinanten $D = f_{xx} f_{yy} - (f_{xy})^2$:

* $D > 0$ og $f_{xx} < 0 \implies$ Lokalt maksimum
* $D > 0$ og $f_{xx} > 0 \implies$ Lokalt minimum
* $D < 0 \implies$ Sadelpunkt

1. **$P_1 = (1, 1)$:** $f_{xx} = -4$, $f_{yy} = 6 \implies D = (-4)(6) - 0 = -24 < 0 \implies$ **Sadelpunkt**

2. **$P_2 = (1, -1)$:** $f_{xx} = -4$, $f_{yy} = -6 \implies D = (-4)(-6) - 0 = 24 > 0$, $f_{xx} < 0 \implies$ **Lokalt maksimum**

3. **$P_3 = \left(-\frac{1}{3}, 1\right)$:** $f_{xx} = 4$, $f_{yy} = 6 \implies D = (4)(6) - 0 = 24 > 0$, $f_{xx} > 0 \implies$ **Lokalt minimum**

4. **$P_4 = \left(-\frac{1}{3}, -1\right)$:** $f_{xx} = 4$, $f_{yy} = -6 \implies D = (4)(-6) - 0 = -24 < 0 \implies$ **Sadelpunkt**



```python
crit_pts = sp.solve([fx, fy], (x, y))
print("Stasjonære punkt:", crit_pts)

D_expr = fxx * fyy - fxy**2

for pt in crit_pts:
    D_val = D_expr.subs({x: pt[0], y: pt[1]})
    fxx_val = fxx.subs({x: pt[0], y: pt[1]})
    
    if D_val < 0:
        kategori = "Sadelpunkt"
    elif D_val > 0 and fxx_val < 0:
        kategori = "Lokalt maksimum"
    elif D_val > 0 and fxx_val > 0:
        kategori = "Lokalt minimum"
    else:
        kategori = "Uavklart"
        
    print(f"Punkt {pt}: D = {D_val}, fxx = {fxx_val} --> {kategori}")

```

## Oppgave 2

Scriptet består av $k$ nøstede løkker:

* Ytre løkke kjører $n_1$ ganger.
* Løkke 2 kjører $n_2$ ganger.
* ...
* Løkke $k$ kjører $n_k$ ganger.

For hver iterasjon av den innerste løkka øker variabelen $k$ med 1.
Det totale antallet iterasjoner blir produktet av antall iterasjoner i hver enkelt løkke:


$$k = \prod_{j=1}^k n_j = n_1 \cdot n_2 \cdot \dots \cdot n_k$$



```python
import sympy as sp

n = sp.IndexedBase('n')
j = sp.symbols('j', integer=True)
k_symbol = sp.Product(n[j], (j, 1, sp.symbols('k')))

print("Sluttverdi for k uttrykt symbolsk:")
display(k_symbol)
```

### a) Grafene til bijektive funksjoner og kardinalitet


Mengdene er $A = \{1, 2, 3\}$ og $B = \{a, b, c\}$.
En bijeksjon er en funksjon som er både injektiv (én-til-én) og surjektiv (på). Siden båda mengdene har 3 elementer, er antall bijektive funksjoner lik antall permutasjoner av 3 elementer:


$$3! = 3 \times 2 \times 1 = 6$$

De 6 bijeksjonene kan skrives som mengder av par $(x, f(x))$:

1. $f_1 = \{(1,a), (2,b), (3,c)\}$
2. $f_2 = \{(1,a), (2,c), (3,b)\}$
3. $f_3 = \{(1,b), (2,a), (3,c)\}$
4. $f_4 = \{(1,b), (2,c), (3,a)\}$
5. $f_5 = \{(1,c), (2,a), (3,b)\}$
6. $f_6 = \{(1,c), (2,b), (3,a)\}$

**Kardinalitet:**
At det finnes minst én bijeksjon mellom $A$ og $B$ definerer per konstruksjon at mengdene har **samme kardinalitet**, det vil si $\vert{}A\vert{} = \vert{}B\vert{} = 3$.


### b) Antall bijektive funksjoner for $\vert{}A\vert{}=\vert{}B\vert{}=7$


For $A = \{1, 2, 3, 4, 5, 6, 7\}$ og $B = \{a, b, c, d, e, f, g\}$ har begge mengder 7 elementer.
Ved **multiplikasjonsprinsippet**:

* Element $1 \in A$ kan avbildes til et hvilket som helst av de 7 elementene i $B$ (7 valg).
* Element $2 \in A$ må avbildes til et ubenyttet element i $B$ (6 valg).
* Element $3 \in A$ har 5 valg.
* ...
* Element $7 \in A$ har 1 valg igjen.

Totalt antall bijektive funksjoner er:


$$7! = 7 \cdot 6 \cdot 5 \cdot 4 \cdot 3 \cdot 2 \cdot 1 = 5040$$



```python
ans_3b = sp.factorial(7)
print("Antall bijektive funksjoner:", ans_3b)
```

### a) Bijeksjon mellom $\mathbb{N}$ og $S = \{1, 6, 11, 16, \dots, 5n+1, \dots\}$

La $\mathbb{N} = \{0, 1, 2, 3, \dots\}$. Definer funksjonen $f: \mathbb{N} \to S$ ved:


$$f(n) = 5n + 1$$

**Bevis for bijeksjon:**

1. **Injektiv:** Anta $f(n_1) = f(n_2)$.

$$5n_1 + 1 = 5n_2 + 1 \implies 5n_1 = 5n_2 \implies n_1 = n_2$$



Dermed er $f$ injektiv.
2. **Surjektiv:** La $y \in S$. Per definisjon av $S$ er $y = 5k + 1$ for en $k \in \mathbb{N}$. Da finnes $n = k \in \mathbb{N}$ slik at $f(n) = 5k + 1 = y$. Dermed er $f$ surjektiv.

Siden $f$ er både injektiv og surjektiv, er den en bijeksjon, og mengdene har samme kardinalitet ($\vert{}S\vert{} = \aleph_0$).



### b) Bijeksjon mellom $(0, 1)$ og $(0, 10)$


Vi definerer funksjonen $g: (0, 1) \to (0, 10)$ ved:


$$g(x) = 10x$$

**Bevis for bijeksjon:**

1. **Veldefinert:** Hvis $0 < x < 1$, så er $0 < 10x < 10$, så $g(x) \in (0, 10)$.
2. **Injektiv:** Anta $g(x_1) = g(x_2) \implies 10x_1 = 10x_2 \implies x_1 = x_2$.
3. **Surjektiv:** La $y \in (0, 10)$. Velg $x = \frac{y}{10}$. Siden $0 < y < 10$, er $0 < \frac{y}{10} < 1$, så $x \in (0, 1)$. Da er $g(x) = 10\left(\frac{y}{10}\right) = y$.

Dermed er $g$ en bijeksjon, og mengdene har samme kardinalitet ($c = 2^{\aleph_0}$).



```python
n1, n2, x1, x2, y = sp.symbols('n1 n2 x1 x2 y', real=True)
fn = 5*n1 + 1
gx = 10*x1

# Verifiserer injektivitet ved å løse fn1 = fn2
print("Løsning f(n1) = f(n2):", sp.solve(5*n1 + 1 - (5*n2 + 1), n1))
print("Løsning g(x1) = g(x2):", sp.solve(10*x1 - 10*x2, x1))

```

### a) Hvor mange strøk på nøyaktig ett emne?


Vi bruker prinsipper fra Venn-diagrammer/inkludering-ekskludering:

1. **Kun Statistikk ($S_{kun}$):**

$$\vert{}S_{kun}\vert{} = \vert{}S\vert{} - \vert{}S \cap M\vert{} - \vert{}S \cap P\vert{} + \vert{}S \cap M \cap P\vert{} = 12 - 2 - 6 + 1 = 5$$


2. **Kun Matematikk ($M_{kun}$):**

$$\vert{}M_{kun}\vert{} = \vert{}M\vert{} - \vert{}S \cap M\vert{} - \vert{}M \cap P\vert{} + \vert{}S \cap M \cap P\vert{} = 5 - 2 - 3 + 1 = 1$$


3. **Kun Programmering ($P_{kun}$):**

$$\vert{}P_{kun}\vert{} = \vert{}P\vert{} - \vert{}P \cap S\vert{} - \vert{}M \cap P\vert{} + \vert{}S \cap M \cap P\vert{} = 8 - 6 - 3 + 1 = 0$$



Totalt antall som strøk på nøyaktig ett emne:


$$\vert{}S_{kun}\vert{} + \vert{}M_{kun}\vert{} + \vert{}P_{kun}\vert{} = 5 + 1 + 0 = 6$$



### b) Hvor mange strøk bare i diskret matematikk?


Som beregnet i del (a) er antall studenter som strøk **bare** diskret matematikk:


$$\vert{}M_{kun}\vert{} = 1$$

### b) Heltallsløsninger til $a+b+c+d+e=500$ der $a,b,c,d,e \ge 10$


Gitt likningen:


$$a + b + c + d + e = 500, \quad \text{med } a,b,c,d,e \ge 10$$

Vi gjør et variabelskifte for å omforme til standard "stjerner og stolper"-problem med non-negative heltall:
La $a' = a - 10$, $b' = b - 10$, $c' = c - 10$, $d' = d - 10$, $e' = e - 10$.
Siden $a,b,c,d,e \ge 10$, vil $a',b',c',d',e' \ge 0$.

Innsetting i likningen gir:


$$(a' + 10) + (b' + 10) + (c' + 10) + (d' + 10) + (e' + 10) = 500$$

$$a' + b' + c' + d' + e' + 50 = 500$$

$$a' + b' + c' + d' + e' = 450$$

Antall løsninger for $k=5$ variabler som summerer seg til $n=450$ med $x_i \ge 0$ er gitt ved formelen:


$$\binom{n + k - 1}{k - 1} = \binom{450 + 5 - 1}{5 - 1} = \binom{454}{4}$$



```python
ans_9b = sp.binomial(454, 4)
print("Antall løsninger:", ans_9b)
```
